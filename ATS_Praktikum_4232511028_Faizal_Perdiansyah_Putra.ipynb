```python
# Soal 1

# a. Simpan tegangan dan arus dalam variabel, lalu hitung daya listrik
tegangan = 220   # satuan Volt
arus = 15        # satuan Ampere
daya = tegangan * arus   # P = V x I

print(f"Daya listrik = {daya} W")

# b. Hitung efisiensi turbin
energi_input = 500
energi_output = 400
efisiensi = (energi_output / energi_input) * 100

print(f"Efisiensi = {efisiensi:.2f}%")
```

**Cara kerja program:**

- `tegangan` dan `arus` adalah variabel yang menyimpan nilai 220 Volt dan 15 Ampere.
- `daya = tegangan * arus` menghitung daya listrik dengan rumus P = V x I, hasilnya 3300.
- f-string (`f"..."`) dipakai untuk menampilkan nilai variabel di dalam teks. Variabel ditulis di dalam kurung kurawal `{}`.
- `efisiensi = (energi_output / energi_input) * 100` menghitung 400 dibagi 500 (0.8), lalu dikali 100 jadi 80.
- `{efisiensi:.2f}` membuat angka tampil dengan 2 angka desimal, jadi hasilnya `80.00`.

```python
# Soal 3

suhu = [92.5, 88.7, 95.1, 102.3, 97.8]

# a. Suhu pada titik ke-3 (index ke-2)
print(f"Suhu titik ke-3 = {suhu[2]} °C")

# b. Tiga pembacaan pertama dengan slicing
print(f"Tiga pembacaan pertama = {suhu[:3]}")

# c. Tambah pembacaan baru, lalu tampilkan jumlah data
suhu.append(99.4)
print(f"Jumlah data setelah penambahan = {len(suhu)}")
```

**Cara kerja program:**

- Index list di Python dimulai dari 0, jadi titik ke-3 diambil dengan `suhu[2]`, hasilnya 95.1.
- `suhu[:3]` adalah slicing untuk mengambil dari awal sampai sebelum index 3, yaitu tiga data pertama.
- `suhu.append(99.4)` menambahkan nilai 99.4 ke akhir list.
- `len(suhu)` menghitung jumlah isi list. Setelah penambahan jumlahnya menjadi 6.

```python
# Soal 5

tekanan_uap = float(input("Masukkan tekanan uap (psi): "))
status_manual_override = True   # ganti True/False untuk pengujian

if tekanan_uap > 150 and status_manual_override == False:
    print("Katup Terbuka")
elif status_manual_override == True:
    print("Katup Mode Manual")
else:
    print("Katup Tertutup")
```

**Cara kerja program:**

- `input()` selalu menghasilkan teks, jadi diubah dengan `float()` supaya bisa dibandingkan sebagai angka.
- `status_manual_override` ditulis langsung di kode sesuai permintaan soal.
- Kondisi `if` pertama memakai operator `and`, jadi baru benar kalau tekanan lebih dari 150 **dan** override bernilai `False`. Hasilnya "Katup Terbuka".
- `elif` dicek kalau kondisi pertama tidak terpenuhi. Kalau override `True`, hasilnya "Katup Mode Manual", berapa pun tekanannya.
- `else` menangani sisanya, yaitu "Katup Tertutup".
- Soal menyebut "lebih dari 150", jadi dipakai `>` bukan `>=`. Karena itu tekanan tepat 150 dengan override `False` hasilnya "Katup Tertutup".

```python
# Soal 7

suhu_saat_ini = 60

while suhu_saat_ini < 85:
    suhu_saat_ini += 5                      # naikkan 5 derajat
    print(f"Suhu saat ini: {suhu_saat_ini} C")

    if suhu_saat_ini == 80:                 # cek tepat 80
        print("WARNING: Mendekati target!")

print("Target 85 C tercapai.")
```

**Cara kerja program:**

- `suhu_saat_ini = 60` adalah suhu awal sebelum perulangan dimulai.
- `while suhu_saat_ini < 85` mengulang kode di dalamnya selama suhu masih di bawah 85. Setelah suhu mencapai 85 perulangan berhenti.
- `suhu_saat_ini += 5` menambah suhu 5 derajat di setiap putaran.
- `print` dijalankan setelah suhu dinaikkan, jadi suhu pertama yang tampil adalah 65.
- `if suhu_saat_ini == 80` mengecek apakah suhu tepat 80. Kalau ya, pesan WARNING dicetak.
- Pesan "Target 85 C tercapai." berada di luar `while`, jadi hanya dicetak sekali setelah perulangan selesai.

```python
# Soal 9

data = ["1.2", "0.9", "ERR", "3.1", "1.8"]
jumlah_tidak_valid = 0

for nomor, teks in enumerate(data, start=1):
    # a. Ubah teks ke float, tangani kalau gagal
    try:
        nilai = float(teks)
    except ValueError:
        print(f"Unit {nomor}: data tidak valid")
        jumlah_tidak_valid += 1
        continue                      # lanjut ke unit berikutnya

    # b. Cek vibrasi tinggi
    if nilai > 2.0:
        print(f"Warning: Unit {nomor} vibrasi tinggi!")

        # c. Cek kritis, lalu hentikan perulangan
        if nilai > 3.0:
            print("KRITIS: pemeriksaan dihentikan")
            break

# d. Tampilkan jumlah data tidak valid
print(f"Jumlah data tidak valid: {jumlah_tidak_valid}")
```

**Cara kerja program:**

- `jumlah_tidak_valid = 0` adalah penghitung data rusak yang dibuat sebelum perulangan.
- `enumerate(data, start=1)` memberi nomor unit yang dimulai dari 1 beserta isi datanya di setiap putaran.
- `try` mencoba mengubah teks menjadi `float`. Untuk `"ERR"` terjadi `ValueError`, lalu `except ValueError` menangkapnya supaya program tidak berhenti. Pesan dicetak dan penghitung bertambah 1.
- `continue` melewati sisa kode pada putaran itu dan lanjut ke unit berikutnya.
- Jika nilai lebih dari 2.0, dicetak Warning vibrasi tinggi.
- Jika nilai lebih dari 3.0, dicetak pesan KRITIS lalu `break` menghentikan seluruh perulangan, jadi unit 5 tidak diperiksa.
- Baris terakhir berada di luar `for`, jadi jumlah data tidak valid ditampilkan sekali setelah perulangan selesai.

```python
# Challenge: Pemantauan Beban Listrik Harian
#
# Masalah:
# Sebuah pabrik kecil mencatat beban listrik (kW) tiap jam. Datanya dibaca
# sebagai teks dan ada sensor yang kadang rusak ("ERR"). Program ini
# mengklasifikasikan beban tiap jam (Normal/TINGGI/KRITIS), mencari beban
# puncak, lalu menghitung total konsumsi (kWh) dan estimasi biaya.
#
# Cara kerja singkat:
# 1. Data dibaca satu per satu dengan for.
# 2. Tiap data diubah ke float dengan try-except. Kalau gagal, dilewati.
# 3. Data valid disimpan ke list, lalu diklasifikasikan dengan if-elif-else.
# 4. Setelah loop, hitung total, rata-rata, dan biaya.

# --- Data dan pengaturan ---
jam_mulai = 6                  # pencatatan mulai jam 06:00
tarif = 1500                   # Rp per kWh
batas_tinggi = 90              # kW, di atas ini dianggap TINGGI
batas_kritis = 120             # kW, di atas ini dianggap KRITIS

data_beban = ["45", "60", "75", "ERR", "95", "110",
              "105", "88", "70", "125", "90", "55"]

beban_valid = []               # list untuk menyimpan data yang valid
jumlah_tinggi = 0
jumlah_kritis = 0
jumlah_error = 0
beban_puncak = 0
jam_puncak = 0

print("=== Laporan Beban Listrik Harian ===")

# --- Proses tiap data ---
for i, teks in enumerate(data_beban):
    jam = jam_mulai + i

    try:
        beban = float(teks)
    except ValueError:
        print(f"Jam {jam:02d}:00 -> data sensor rusak, dilewati")
        jumlah_error += 1
        continue

    beban_valid.append(beban)

    # klasifikasi beban
    if beban > batas_kritis:
        status = "KRITIS"
        jumlah_kritis += 1
    elif beban > batas_tinggi:
        status = "TINGGI"
        jumlah_tinggi += 1
    else:
        status = "Normal"

    # cari beban puncak
    if beban > beban_puncak:
        beban_puncak = beban
        jam_puncak = jam

    print(f"Jam {jam:02d}:00 -> {beban:.1f} kW ({status})")

# --- Ringkasan ---
print("\n=== Ringkasan ===")

if len(beban_valid) > 0:
    total_kwh = sum(beban_valid)           # tiap data mewakili 1 jam
    rata_rata = total_kwh / len(beban_valid)
    biaya = total_kwh * tarif

    print(f"Total konsumsi    : {total_kwh:.1f} kWh")
    print(f"Rata-rata beban   : {rata_rata:.2f} kW")
    print(f"Beban puncak      : {beban_puncak:.1f} kW pada jam {jam_puncak:02d}:00")
    print(f"Jam beban tinggi  : {jumlah_tinggi}")
    print(f"Jam beban kritis  : {jumlah_kritis}")
    print(f"Data sensor rusak : {jumlah_error}")
    print(f"Estimasi biaya    : Rp {biaya:,.0f}")

    if jumlah_kritis > 0:
        print("Evaluasi: ada jam kritis, beban perlu dikurangi saat jam puncak!")
    elif jumlah_tinggi > 0:
        print("Evaluasi: ada jam beban tinggi, perlu dipantau.")
    else:
        print("Evaluasi: beban listrik aman.")
else:
    print("Tidak ada data valid yang bisa dihitung.")
```

**Cara kerja program:**

- `data_beban` berisi data beban per jam dalam bentuk teks, dan `beban_valid` adalah list kosong yang diisi dengan `append` setiap ada data yang valid.
- `for` dengan `enumerate` membaca tiap data, dan `jam = jam_mulai + i` menghitung jam pencatatan dari index.
- `try-except ValueError` menangani data "ERR" supaya program tidak berhenti. Data rusak dihitung lalu dilewati dengan `continue`.
- `if-elif-else` mengklasifikasikan beban: di atas 120 kW KRITIS, di atas 90 kW TINGGI, selain itu Normal.
- Beban puncak dicari dengan membandingkan setiap nilai dengan rekor tertinggi sebelumnya.
- Setelah perulangan, `sum()` dan `len()` dipakai untuk menghitung total, rata-rata, dan biaya. `if len(beban_valid) > 0` mencegah pembagian dengan nol kalau semua data rusak.
- Format `:02d` membuat jam tampil dua digit, `:.2f` memberi 2 angka desimal, dan `:,.0f` menambah pemisah ribuan pada biaya.
