```python
# Data 24 jam: suhu bantalan turbin

# Suhu bantalan turbin (°C), satu nilai per jam
suhu = [
    61, 60, 60, 59, 59, 60,   # pukul 00.00–05.00
    63, 66, 69, 71, 72, 73,   # pukul 06.00–11.00
    74, 74, 75, 74, 73, 72,   # pukul 12.00–17.00
    70, 68, 66, 64, 63, 62,   # pukul 18.00–23.00
]

print("Jumlah data :", len(suhu))         # 24 elemen: satu per jam
print("Pukul 00.00 :", suhu[0], "°C")     # indeks 0 = elemen pertama
print("Pukul 14.00 :", suhu[14], "°C")    # indeks = jam, karena mulai pukul 00.00
print("Pukul 23.00 :", suhu[-1], "°C")    # -1 = elemen terakhir
```

```python
# Memotong per periode dengan slicing

suhu = [
    61, 60, 60, 59, 59, 60,   # pukul 00.00–05.00
    63, 66, 69, 71, 72, 73,   # pukul 06.00–11.00
    74, 74, 75, 74, 73, 72,   # pukul 12.00–17.00
    70, 68, 66, 64, 63, 62,   # pukul 18.00–23.00
]

malam = suhu[0:6]     # pukul 00.00–05.00 (indeks 6 tidak ikut)
pagi = suhu[6:12]     # pukul 06.00–11.00
siang = suhu[12:18]   # pukul 12.00–17.00
sore = suhu[18:]      # akhir dikosongkan = sampai elemen terakhir

print("Pagi       :", pagi)
print("Siang      :", siang)
print("Tiap 3 jam :", suhu[::3])       # langkah 3: indeks 0, 3, 6, ..., 21
print("6 jam akhir:", suhu[-6:])       # 6 elemen terakhir
print("Dibalik    :", suhu[::-1][:4])  # [::-1] membalik, lalu [:4] ambil 4 pertama
print("Panjang    :", len(malam), len(pagi), len(siang), len(sore))
```

# **Perulangan dan percabangan dengan array**

```python
# Menelusuri list: dua cara

skor = [80, 65, 90, 75, 70]

print("range(5) menghasilkan:", list(range(5)))

# 1) for-each: ambil setiap NILAI secara bergiliran
for nilai in skor:
    print("skor:", nilai)

# 2) berdasarkan posisi: i adalah INDEKS, skor[i] adalah nilainya
for i in range(len(skor)):
    print("indeks", i, "berisi", skor[i])
```

```python
# Pola akumulator: total, rata-rata, hitungan

skor = [80, 65, 90, 75, 70]

total = 0                    # akumulator: mulai dari nol
jumlah_lulus = 0             # penghitung = akumulator yang menambah 1
for nilai in skor:
    total = total + nilai    # tambahkan setiap elemen
    if nilai >= 75:
        jumlah_lulus = jumlah_lulus + 1
    print("total sementara:", total)

rata_rata = total / len(skor)
print("Total     :", total)
print("Rata-rata :", rata_rata)
print("Skor >= 75:", jumlah_lulus, "dari", len(skor))
```

```python
# Nilai tertinggi dan terendah

skor = [80, 65, 90, 75, 70]

tertinggi = skor[0]          # terbaik sejauh ini (anggap yang pertama)
terendah = skor[0]
for nilai in skor:
    if nilai > tertinggi:
        tertinggi = nilai
    if nilai < terendah:
        terendah = nilai

print("Tertinggi:", tertinggi)
print("Terendah :", terendah)
print("Cek fungsi bawaan:", max(skor), min(skor))
```

```python
# Bukan hanya nilainya, tetapi posisinya
# max() memberi tahu suhu tertinggi tetapi tidak hari ke berapa. Untuk itu, ingat indeks elemen terbaik, bukan nilainya; nilainya kemudian adalah suhu[pos_maks].

suhu = [28, 31, 30, 33, 29]      # suhu (°C), hari ke-1 sampai ke-5

pos_maks = 0                      # posisi terbaik sejauh ini
for i in range(len(suhu)):
    if suhu[i] > suhu[pos_maks]:
        pos_maks = i              # simpan POSISI, bukan hanya nilai

print("Suhu tertinggi:", suhu[pos_maks], "°C")
print("Ada di indeks", pos_maks, "= hari ke-", pos_maks + 1)
```

```python
# Pencarian linear: berhenti di kecocokan pertama

suhu = [28, 31, 30, 33, 29]
target = 32                       # hari pertama yang lebih panas dari 32 °C?

ketemu = None                     # None = "belum ketemu"
cek = 0
for i in range(len(suhu)):
    cek = cek + 1
    if suhu[i] > target:
        ketemu = i
        break                     # berhenti di kecocokan pertama

print("Hari pertama di atas", target, "°C: indeks", ketemu)
print("Elemen yang diperiksa:", cek, "dari", len(suhu))

# Jalan pintas yang berguna
print(33 in suhu)                 # apakah 33 ada di dalam list?
print(suhu.index(33))             # posisi 33 yang pertama (error jika tidak ada)
```

```python
# Penyaringan: list baru berisi elemen yang cocok

suhu = [28, 31, 30, 33, 29]

panas = []                        # list baru, awalnya kosong
for t in suhu:
    if t >= 30:
        panas.append(t)
print("Nilai panas :", panas)
print("Jumlahnya   :", len(panas))

# Untuk menyimpan nomor hari, bukan nilainya, telusuri berdasarkan posisi
hari_panas = []
for i in range(len(suhu)):
    if suhu[i] >= 30:
        hari_panas.append(i + 1)
print("Hari panas  :", hari_panas)

# Opsional: penyaringan yang sama dengan list comprehension (perulangan ringkas)
panas_lagi = [t for t in suhu if t >= 30]
print("Hasil sama?", panas_lagi == panas)
```

```python
# Tetangga: perubahan harian dan rata-rata bergerak

suhu = [28, 31, 30, 33, 29, 27, 32]

# Perubahan harian: mulai dari indeks 1 karena kita melihat i - 1
for i in range(1, len(suhu)):
    perubahan = suhu[i] - suhu[i - 1]
    print(f"hari {i + 1}: {suhu[i]} °C, perubahan {perubahan:+d}")

# Rata-rata bergerak 3 hari: jendela = 3 nilai yang berakhir di indeks i
rata_3 = []
for i in range(2, len(suhu)):
    jendela = suhu[i - 2 : i + 1]
    rata_3.append(round(sum(jendela) / 3, 1))

print("Rata-rata 3 hari:", rata_3)
print(len(suhu), "nilai ->", len(rata_3), "rata-rata")
```

```python
# Menampilkan Hari dengan Rata-rata Bergerak 7 Hari di Bawah 97% Rata-rata Bulanan

energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564,
    5552, 6100, 6083, 6097, 6089, 6066, 5533,
    5566, 6097, 3592, 6059, 6100, 6098, 5565,
    5571, 6103, 6135, 6065, 6071, 6086, 5577,
    5575, 6062
]

rata_bulanan = sum(energi) / len(energi)
batas = 0.97 * rata_bulanan

print("Batas 97%:", batas)

for i in range(6, len(energi)):
    rata_7_hari = sum(energi[i-6:i+1]) / 7

    if rata_7_hari < batas:
        print(
            "Hari", i + 1,
            "Rata-rata 7 hari:", round(rata_7_hari, 2), "MWh"
        )
```

```python
# Menghitung Total Energi Per Minggu dan Sisa Hari

energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564,
    5552, 6100, 6083, 6097, 6089, 6066, 5533,
    5566, 6097, 3592, 6059, 6100, 6098, 5565,
    5571, 6103, 6135, 6065, 6071, 6086, 5577,
    5575, 6062
]

for i in range(0, 28, 7):
    minggu = energi[i:i+7]
    nomor_minggu = (i // 7) + 1

    print(
        "Minggu", nomor_minggu,
        ":", sum(minggu), "MWh"
    )

sisa_hari = energi[28:]
print("Sisa hari (29-30):", sum(sisa_hari), "MWh")
```
