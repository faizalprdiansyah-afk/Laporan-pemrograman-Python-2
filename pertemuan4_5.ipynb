```python
# Soal 1 - Kalkulator faktorial 1 sampai 5
n = int(input("Masukkan bilangan bulat (1-5): "))

if n < 1 or n > 5:
    print("Input tidak valid. Masukkan bilangan antara 1 sampai 5.")
elif n == 1:
    hasil = 1
    print(f"{n}! = {hasil}")
elif n == 2:
    hasil = 2 * 1
    print(f"{n}! = {hasil}")
elif n == 3:
    hasil = 3 * 2 * 1
    print(f"{n}! = {hasil}")
elif n == 4:
    hasil = 4 * 3 * 2 * 1
    print(f"{n}! = {hasil}")
else:
    hasil = 5 * 4 * 3 * 2 * 1
    print(f"{n}! = {hasil}")
```

```python
# Soal 2 — Klasifikasi kualitas daya
faktor_daya = float(input("Masukkan faktor daya (0.0 - 1.0): "))
efisiensi   = float(input("Masukkan efisiensi distribusi (%): "))

skor = faktor_daya * efisiensi

print(f"Skor kualitas daya: {skor:.2f}")

if skor >= 85 and faktor_daya >= 0.90:
    klasifikasi = "PRIMA"
elif skor >= 70 and faktor_daya >= 0.80:
    klasifikasi = "BAIK"
elif skor >= 50 or faktor_daya >= 0.75:
    klasifikasi = "CUKUP"
else:
    klasifikasi = "BURUK"

print(f"Klasifikasi: {klasifikasi}")
```

```python
# Soal 3 — Rekomendasi mode operasi PLTS
iradiasi = float(input("Masukkan iradiasi matahari (W/m²): "))
soc      = float(input("Masukkan SoC baterai (%): "))

daya_panel = iradiasi * 0.18 * 10

print(f"Estimasi daya panel: {daya_panel:.1f} W")

if iradiasi > 600 and soc < 80:
    mode = "CHARGING"
elif iradiasi > 600 and soc >= 80:
    mode = "GRID-EXPORT"
elif iradiasi > 200 and soc >= 20:
    mode = "NORMAL"
elif iradiasi <= 200 and soc >= 50:
    mode = "BATTERY-ONLY"
else:
    mode = "EMERGENCY"

print(f"Mode: {mode}")
```
